## 1.2. FUN

```python
nums = [1, 2, 3, 3, 2, 2, 9]
unique = []

i = 0
while i < len(nums):
    value = nums[i]
    print(i)
    if value not in unique:
        unique.append(value)
    i += 1

# sort the result
unique.sort()
print(unique)
```
