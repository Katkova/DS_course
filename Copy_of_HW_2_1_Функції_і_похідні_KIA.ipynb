### Домашні завдання на відпрацювання поняття функції та її похідної

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Визначення функції та її похідної

x = sp.symbols('x') # Визначаємо символічну змінну
f = 10 * x - 5

# Похідна
f_derivative = sp.diff(f, x)

print(f"Функція: f(x) = {f}")
print(f"Похідна: f'(x) = {f_derivative}")
```

```python
# Визначення функції та її похідної
x_vals = np.linspace(0, 10, 100)
f_point = 10 * x_vals - 5
df = np.full_like(x_vals, 10)  # 100 точок зі значенням 10

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x_vals, f_point, label='f(x) = 10x - 5', color='blue')
plt.plot(x_vals, df, label="f'(x) = 10", color='green')

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("Години")
plt.ylabel("Оцінка")
plt.legend()
plt.grid(True)

plt.show()
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x') # Визначаємо символічну змінну

# Визначення функцій
f_sin = sp.sin(x)
f_ln = sp.ln(x)

# Похідні
df_sin = sp.diff(f_sin, x)
df_ln = sp.diff(f_ln, x)

print(f"f(x) = sin(x) похідна = {df_sin}")
print(f"f(x) = ln(x) похідна = {df_ln}")
```

```python
# Числові значення
x_vals = np.linspace(0.2, 10, 100)

# Графік sin(x)
plt.figure(figsize=(10, 5))
plt.plot(x_vals, np.sin(x_vals), label='f(x) = sin(x)', color='blue')
plt.plot(x_vals, np.cos(x_vals), label="f'(x) = cos(x)", color='green')
plt.title("sin(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()

# Графік ln(x)
plt.figure(figsize=(10, 5))
plt.plot(x_vals, np.log(x_vals), label='f(x) = ln(x)', color='blue')
plt.plot(x_vals, 1/x_vals, label="f'(x) = 1/x", color='green')
plt.title("ln(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.Symbol('x')

# Визначення функції
f3 = x**2 - 5*x + 1

# Похідна
df = sp.diff(f3, x)
print(f"Функція:  f(x) = {f3}")
print(f"Похідна: f'(x) = {df}")
```

```python
# Числові значення
x_vals = np.linspace(0, 100, 100)

# Графік
plt.figure(figsize=(10, 5))
plt.plot(x_vals, x_vals**2 - 5*x_vals + 1, label='f(x) = x² - 5x + 1', color='blue')
plt.plot(x_vals, 2*x_vals - 5, label="f'(x) = 2x - 5", color='green')
plt.title("Залежність загальних витрат компанії від обсягу виробництва")
plt.xlabel("Кількість одиниць")
plt.ylabel("Витрати")
plt.legend()
# plt.grid(True)

plt.show()
```

```python
# Трошки економічної теорії з університету і допомоги АІ

# Знаходимо точку беззбитковості де f'(x) = 0
x_zero = sp.solve(df, x)
print(f"Точка беззбитковості: x = {x_zero[0]}")
print(f"Витрати в цій точці: f({x_zero[0]}) = {f.subs(x, x_zero[0])}")

# Числові значення
x_vals = np.linspace(0, 100, 100)

# Графік
plt.figure(figsize=(10, 5))
plt.plot(x_vals, x_vals**2 - 5*x_vals + 1, label='f(x) = x² - 5x + 1', color='blue')
plt.plot(x_vals, 2*x_vals - 5, label="f'(x) = 2x - 5", color='green')

# Позначаємо точку беззбитковості
x_point = float(x_zero[0])
f_point = float(f.subs(x, x_point))
plt.scatter(x_point, f_point, color='red', zorder=5)
plt.annotate(f'Точка беззбитковості\nx={x_point}, f(x)={f_point}',
             xy=(x_point, f_point),
             xytext=(x_point + 5, f_point + 500),
             arrowprops=dict(arrowstyle='->'))

plt.title("Витрати компанії та швидкість їх зростання")
plt.xlabel("Кількість одиниць (x)")
plt.ylabel("Витрати")
plt.legend()
plt.grid(True)
plt.show()
```
